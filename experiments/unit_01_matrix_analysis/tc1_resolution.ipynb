```python
import numpy as np

registration = 318375

# Seed with my number registration
np.random.seed(registration)

# 1. Generation of random matrix A
A = np.random.choice([-1, 0, 1], size=(8, 16))

# 2. Manual calculation (via theoretical definitions) for numerical double-checking
norm_1_manual = np.max(np.sum(np.abs(A), axis=0))
norm_inf_manual = np.max(np.sum(np.abs(A), axis=1))
norm_F_manual = np.sqrt(np.sum(A**2))

# Calculation of norm-2 via SVD decomposition (square root of the largest eigenvalue of A^T A),
# where, ||A||_2 = max singular value
U, S, Vt = np.linalg.svd(A)
norm_2_manual = S[0]

# 3. Calculation via standard linear algebra library (Numpy via np.linalg.norm)
norm_1_lib = np.linalg.norm(A, ord=1)
norm_2_lib = np.linalg.norm(A, ord=2)
norm_inf_lib = np.linalg.norm(A, ord=np.inf)
norm_F_lib = np.linalg.norm(A, ord='fro')

# 4. Strict formatting with 6 significant figures
print(f"Matrix A Shape: {A.shape}")
print(f"||A||_1   = {norm_1_lib:.6g}")  # Column Maximum Sum Standard
print(f"||A||_2   = {norm_2_lib:.6g}")  # Maximum singular value of matrix A, via SVD decomposition
print(f"||A||_inf = {norm_inf_lib:.6g}")  # Line Maximum Sum Standard
print(f"||A||_F   = {norm_F_lib:.6g}")  # Frobenius Norm
```

## Question 1-c:

```python
import numpy as np

# Same seed and matrix from Item (a)
np.random.seed(registration)
A = np.random.choice([-1, 0, 1], size=(8, 16))

# Full SVD calculation
U, S, Vt = np.linalg.svd(A, full_matrices=True)

# Construction of the Sigma matrix (8x16) from the 1D array of singular values
Sigma = np.zeros((8, 16))
np.fill_diagonal(Sigma, S)

# Recovering the V matrix from Vt
V = Vt.T

# --- Helper function to generate LaTeX code ---
def matrix_to_latex(matrix, name, precision=4):
    """Converts a numpy array to a LaTeX pmatrix string."""
    latex_str = f"\\textbf{{Matrix }} ${name} \\in \\mathbb{{R}}^{{{matrix.shape[0]} \\times {matrix.shape[1]}}}$: \n"
    latex_str += "\\begin{equation*}\n"
    # If the matrix is too large (like 16x16), we use resizebox to fit on the page
    if matrix.shape[1] > 8:
        latex_str += "\\resizebox{\\textwidth}{!}{$\n"
    
    latex_str += "\\begin{bmatrix}\n"
    for row in matrix:
        # Format each number with the required precision
        formatted_row = [f"{val:.{precision}f}" for val in row]
        latex_str += " & ".join(formatted_row) + " \\\\\n"
    latex_str += "\\end{bmatrix}\n"
    
    if matrix.shape[1] > 8:
        latex_str += "$}\n"
    latex_str += "\\end{equation*}\n"
    return latex_str

# Outputs to terminal and writes to LaTeX file
output_lines = [
    "% --- Lines for Overleaf ---",
    matrix_to_latex(U, "U", precision=6),
    matrix_to_latex(Sigma, "\\Sigma", precision=6),
    matrix_to_latex(V, "V", precision=6)
]

for line in output_lines:
    print(line)

# Save to file
with open("question1c_answer.tex", "w") as f:
    for line in output_lines:
        f.write(line + "\n")
```

## Question 1-d:

```python
import numpy as np

# (Assuming A, U, Sigma and V are already loaded from the previous item)
np.random.seed(registration)
A = np.random.choice([-1, 0, 1], size=(8, 16))
U, S, Vt = np.linalg.svd(A, full_matrices=True)
Sigma = np.zeros((8, 16))
np.fill_diagonal(Sigma, S)
V = Vt.T

# 1. Matrix reconstruction from SVD
# Note: V.T is the transpose of V. In Python, Vt is already V^T.
A_recon = U @ Sigma @ Vt

# 2. Error Matrix
E = A - A_recon

# 3. Calculation of error norms
norm_F_error = np.linalg.norm(E, ord='fro')
norm_2_error = np.linalg.norm(E, ord=2)

print(f"||A - A_recon||_F = {norm_F_error:.6g}")
print(f"||A - A_recon||_2 = {norm_2_error:.6g}")

# 4. Empirical verification of the theoretical inequality
ratio = norm_F_error / norm_2_error
print(f"Ratio (Frobenius / Norm-2): {ratio:.6f} <= sqrt(8) (2.828427)")

# Outputs to terminal and writes to LaTeX file
output_lines = [
    "% --- Lines for Overleaf ---",
    "\\begin{itemize}",
    f"    \\item $\\|A - A_{{recon}}\\|_F = {norm_F_error:.6g}$",
    f"    \\item $\\|A - A_{{recon}}\\|_2 = {norm_2_error:.6g}$",
    f"    \\item Ratio (Frobenius / Norm-2): ${ratio:.6f} \\leq \\sqrt{{8}} \\approx 2.828427$",
    "\\end{itemize}"
]

with open("question1d_answer.tex", "w") as f:
    for line in output_lines:
        f.write(line + "\n")
```

## Question 1-e:

```python
import numpy as np

# Identical setup to the previous items
np.random.seed(registration)
A = np.random.choice([-1, 0, 1], size=(8, 16))
U, S, Vt = np.linalg.svd(A, full_matrices=False)

# Choice of rank k for the approximation
k = 4

# Truncation of matrices (Rank k Approximation)
Uk = U[:, :k]
Sk = np.diag(S[:k])
Vk_T = Vt[:k, :]

# Rank k reconstruction
A_k = Uk @ Sk @ Vk_T

# Error Matrix
E_k = A - A_k

# 1. Computational Calculation of Error Norms
norm_2_error = np.linalg.norm(E_k, ord=2)
norm_F_error = np.linalg.norm(E_k, ord='fro')

# 2. Theoretical Validation (Eckart-Young-Mirsky Theorem)
# Remembering that Python indices start at 0, so S[k] is the (k+1)-th value
theo_norm_2 = S[k]
theo_norm_F = np.sqrt(np.sum(S[k:]**2))

print(f"Rank k chosen: {k}")
print(f"||A - A_k||_2 Computational = {norm_2_error:.6g} (Theoretical Validation = {theo_norm_2:.6g})")
print(f"||A - A_k||_F Computational = {norm_F_error:.6g} (Theoretical Validation = {theo_norm_F:.6g})")

# Outputs to terminal and writes to LaTeX file
output_lines = [
    "% --- Lines for Overleaf ---",
    "\\begin{itemize}",
    f"    \\item Rank $k$ chosen: {k}",
    f"    \\item $\\|A - A_k\\|_2$ Computational = ${norm_2_error:.6g}$ (Theoretical Validation = ${theo_norm_2:.6g}$)",
    f"    \\item $\\|A - A_k\\|_F$ Computational = ${norm_F_error:.6g}$ (Theoretical Validation = ${theo_norm_F:.6g}$)",
    "\\end{itemize}"
]

with open("question1e_answer.tex", "w") as f:
    for line in output_lines:
        f.write(line + "\n")
```
