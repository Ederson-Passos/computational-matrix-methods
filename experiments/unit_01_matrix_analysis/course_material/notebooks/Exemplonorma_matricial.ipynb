```python
import numpy as np

# Definindo a matriz
A = np.array([
    [-3,  5,  7],
    [ 2,  6,  4],
    [ 0, -2,  8]
])

# Norma 1 (máximo das somas das colunas)
norma_1 = np.linalg.norm(A, 1)

# Norma 2 (maior valor singular)
norma_2 = np.linalg.norm(A, 2)

# Norma infinito (máximo das somas das linhas)
norma_inf = np.linalg.norm(A, np.inf)

norma_f = np.linalg.norm(A) #Frobenius

print("Norma 1        =", norma_1)
print("Norma 2        =", norma_2)
print("Norma infinito =", norma_inf)
print("Norma Frobenius =", norma_f)
```
