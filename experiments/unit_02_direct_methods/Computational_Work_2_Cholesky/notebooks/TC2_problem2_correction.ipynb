# Problem 2 Correction
This notebook implements the corrected finite-difference scheme for the steady-state heat equation.

```
import numpy as np
import matplotlib.pyplot as plt

k = 0.0847
f = 0.6000
h = 0.1
N = 11
x = np.linspace(0, 1, N)
```

## Analytical Solution

```
T_exact = (f / (2*k)) * x * (1 - x)
```

## Corrected Finite-Difference Matrix

```
M_corr = np.zeros((N, N))
s_corr = np.zeros(N)
coef = 1/h**2
for i in range(1, N-1):
    M_corr[i, i-1] = -coef
    M_corr[i, i] = 2*coef
    M_corr[i, i+1] = -coef
    s_corr[i] = f / k
M_corr[0, 0] = 1
s_corr[0] = 0
M_corr[-1, -1] = 1
s_corr[-1] = 0

T_corr = np.linalg.solve(M_corr, s_corr)
```

## Comparison Plot

```
plt.figure(figsize=(8,5))
plt.plot(x, T_exact, 'k-', label='Exact')
plt.plot(x, T_corr, 'b--o', label='Corrected')
plt.legend()
plt.grid()
plt.show()
```
